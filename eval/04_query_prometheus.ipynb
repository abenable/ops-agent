# Single-tool eval: `query_prometheus`

Runs the real production agent (`agent.agent.get_agent()`) restricted to `query_prometheus` **plus** its metric/label discovery helpers (`list_metric_names`, `get_label_values`) from `agent.tools.prometheus`. Everything outside this backend is still unavailable for this call — same system prompt, middleware, model, and checkpointer the Discord bot runs, just a narrower tool set. This measures how far the deployed agent gets on this backend alone.


```python
from eval_transcript import run_agent_eval
from agent.tools.prometheus import get_label_values, list_metric_names, query_prometheus
```

```python
TOOLS = [list_metric_names, get_label_values, query_prometheus]
```

### The loop

Every run below goes through `eval_transcript.run_agent_eval`, which drives the real production agent (`agent.agent.get_agent()`) — the same system prompt, middleware, model, and checkpointer the Discord bot runs — restricted to `TOOLS` for this call. It prints each tool call and dumps the full transcript to `eval/logs/<timestamp>__<label>.json`.


## In-scope question

A question the bound tools (`list_metric_names`, `get_label_values`, `query_prometheus`) together should be able to answer.


```python
answer = await run_agent_eval(
    "04_query_prometheus:live-inscope",
    "What is the current value of the 'up' metric?",
    thread_id="04_query_prometheus:live-inscope",
    tools=TOOLS,
)
print(answer)
```

## Out-of-scope question

A question that needs evidence outside this backend. Watch whether the model admits the limit or hallucinates past it.


```python
answer = await run_agent_eval(
    "04_query_prometheus:live-outofscope",
    "How has CPU usage trended over the last 30 minutes?",
    thread_id="04_query_prometheus:live-outofscope",
    tools=TOOLS,
)
print(answer)
```
